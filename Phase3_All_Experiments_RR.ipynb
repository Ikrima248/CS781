## Experiment selection


```python
import torch
from phase3_data import read,write,download_data,prepare_cache,RRDataset,LABELS
from phase3_model import load_original,parameter_counts
from phase3_workflow import plan,launch,collect
from phase2_io import resolve_checkpoints,get_hf_token,checkpoint_candidates
EXPERIMENTS=plan(INCLUDE_MODEL_D,RUN_CONTEXT_ABLATION,ONLY_EXPERIMENTS,RUN_FULL_REFERENCE,RUN_WIDE_LORA)
FAMILIES=sorted({e["family"] for e in EXPERIMENTS})
if any(e["method"]=="kd" for e in EXPERIMENTS):FAMILIES=sorted(set(FAMILIES)|{"C"})
GPU_COUNT=min(MAX_GPUS,torch.cuda.device_count())
print(json.dumps(EXPERIMENTS,indent=2))
print("Visible GPUs:",torch.cuda.device_count(), "Using:",GPU_COUNT)
from importlib.metadata import version
write(ROOT/"runtime_environment.json",{"python":sys.version,"packages":{p:version(p) for p in ("torch","numpy","tokenizers","huggingface_hub")},
    "cuda":torch.version.cuda,"gpus":[torch.cuda.get_device_name(i) for i in range(torch.cuda.device_count())]})
```

## Original checkpoints


```python
def resolve_d_path():
    if MODEL_D_CHECKPOINT and checkpoint_candidates(MODEL_D_CHECKPOINT):return MODEL_D_CHECKPOINT
    if MODEL_D_HF_REPO_ID:
        from huggingface_hub import HfApi,hf_hub_download
        token=get_hf_token(required=True);api=HfApi(token=token)
        revision=api.model_info(MODEL_D_HF_REPO_ID).sha
        names=[n for n in api.list_repo_files(MODEL_D_HF_REPO_ID,revision=revision) if n.endswith(".pt")]
        filename=MODEL_D_HF_FILENAME
        if not filename:
            if len(names)!=1:raise ValueError("Set MODEL_D_HF_FILENAME to the verified compact export; found "+str(names))
            filename=names[0]
        return hf_hub_download(MODEL_D_HF_REPO_ID,filename,revision=revision,token=token)
    raise FileNotFoundError("D checkpoint missing. Set MODEL_D_CHECKPOINT to your attached original .pt/directory, configure its verified HF backup, or set INCLUDE_MODEL_D=False for a clearly incomplete ABC run.")

if RUN_WORKFLOW:
    if GPU_COUNT<1:raise RuntimeError("Enable a GPU before downloading/restoring real checkpoints.")
    overrides=dict(CHECKPOINT_OVERRIDES)
    old_manifest=ROOT/"phase1_manifest.json"
    if old_manifest.exists():
        for family,info in read(old_manifest).items():
            if Path(info["path"]).is_file():overrides.setdefault(family,info["path"])
    if "D" in FAMILIES:overrides["D"]=resolve_d_path()
    foundations=resolve_checkpoints(TOKENIZER_DIR,ROOT/"phase1_weights",overrides=overrides,
        allow_download=ALLOW_PHASE1_DOWNLOADS,families=FAMILIES)
    # Preserve other verified families when running a subset later.
    phase1=read(old_manifest) if old_manifest.exists() else {}
    for family,info in foundations.items():
        model=load_original(info)
        info["original_lm_parameters"]=parameter_counts(model)["total"]
        limit=model.cfg.get("source_length",model.cfg.get("context_length"))
        if math.ceil(MAX_TOKENS/8)*8>limit:raise ValueError("MAX_TOKENS exceeds "+family+" original position capacity.")
        print(family,info["path"],info["original_lm_parameters"],"parameters")
        phase1[family]=info;del model;gc.collect()
    write(old_manifest,phase1)
else:
    print("Real checkpoint preflight skipped.")
```
